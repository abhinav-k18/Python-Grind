```python
#Create a function that returns only orders with net_amount_inr > 500000.

import pandas as pd 

df = pd.read_csv("data/monitor_sales.csv")

def high_value_order(dataframe):
    return list(
        filter(lambda x:x["net_amount_inr"] > 500000, dataframe.to_dict("records"))
    )

orderabove5l = high_value_order(df)

for order in orderabove5l[:5]:
    print(order["order_id"], order["net_amount_inr"])
```

```python
#Write a function that counts the number of orders per region.

import pandas as pd 

df = pd.read_csv("data/monitor_sales.csv")

region_count = df.groupby("region")["order_id"].count()
print(region_count)
```

```python
#Write a function that computes the average discount percentage for each brand.

import pandas as pd 

df = pd.read_csv("data/monitor_sales.csv")

AvgDisPer = df.groupby("brand")["discount_pct"].mean()
print(AvgDisPer)

```

```python
#Write a function that returns the top N customers by total net amount.

import pandas as pd 

df = pd.read_csv("data/monitor_sales.csv")

sortvalues = df.sort_values(by = "net_amount_inr", ascending = False)

max_inr = sortvalues.iloc[:5]

print("Max Net Amount:", max_inr["net_amount_inr"])
print("Customer Name:", max_inr["customer_name"])
```

```python
# filter orders where discount_pct > 10 

import pandas as pd 

df = pd.read_csv("data/monitor_sales.csv")

def DisPct(dataframe):
    return list(
        filter(lambda x :x["discount_pct"] > 10,dataframe.to_dict("records"))
    )

results = DisPct(df)

for values in results:
    print(values["order_id"], values["discount_pct"])

```
