```python
# q6. [medium] find all orders that are either "Cancelled" or "Returned", but exclude orders with a sales_amount below ₹10,000.

import pandas as pd

df = pd.read_csv("data/monitor_sales.csv")

def OrderTypes(dataframe):
    return list (
        filter( 
            lambda x:(x["order_status"] in ["Cancelled" , "Returned"]) and x["net_amount_inr"] >=10000, 
        dataframe.to_dict("records")
        
    )
)
results = OrderTypes(df)

for values in results:
    print(values["order_id"],values["order_status"],values["category"],values["panel_type"])
```
